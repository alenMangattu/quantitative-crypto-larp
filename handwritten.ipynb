```python
import numpy as np
import pandas as pd
import matplotlib as plt
from matplotlib import legend
```

```python
def gbm(n_years=10, mu = 0.07, sigma = 0.15, steps_per_year = 12):
    dt = 1 / steps_per_year
    n_steps = n_years * steps_per_year
    xi = np.random.normal(size=n_steps)
    rets = ((1+mu)**dt)-1 + sigma * np.sqrt(dt) * xi
    return pd.DataFrame(rets) 
```

```python
ret = gbm()

(1+ret).cumprod().plot()
```

```python
def gbm(n_years=10, n_senarios=50, mu = 0.07, sigma = 0.15, steps_per_year = 12, s_0 = 100):
    dt = 1 / steps_per_year
    n_steps = n_years * steps_per_year
    xi = np.random.normal(size=(n_steps, n_senarios))
    rets = ((1+mu)**dt)-1 + sigma * np.sqrt(dt) * xi

    rets = pd.DataFrame(rets)
    return s_0*(1+rets).cumprod()
```

```python
gbm().plot(legend=False, figsize=(12, 6))
```
